# correlation coeffcient

<img style="display: block;-webkit-user-select: none;cursor: zoom-in;background-color: hsl(0, 0%, 90%);transition: background-color 300ms;" src="https://imgs.search.brave.com/TTscP02KLdjYjrFp3hzbYy3iqkcpjmBSneHLveUymRE/rs:fit:860:0:0:0/g:ce/aHR0cHM6Ly9jZG4x/LmJ5anVzLmNvbS93/cC1jb250ZW50L3Vw/bG9hZHMvMjAxOS8w/OS9QZWFyc29uLUNv/ZWZmaWNpZW50LWZv/cm11bGEucG5n" width="400" height="73">

```python
arr1=np.array(x)
meanx=np.mean(arr1)
arr11=arr1-meanx
arr2=np.array(y)
meany=np.mean(arr2)
arr22=arr2-meany
arr3=arr11*arr22
NR=np.sum(arr3)
a=np.sum(np.square(arr11))
b=np.sum(np.square(arr22))
DR=np.sqrt(a*b)
NR/DR
```
