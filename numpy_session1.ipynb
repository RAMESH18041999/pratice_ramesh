# speed test

```python
import time
start=time.time()
l1=list(range(1,1000000))
l2=list(range(1,1000000))
l=[i*j for i,j in zip(l1,l2)]
end=time.time()
print(f"time taken {end-start}")
```

```python
import time
start=time.time()
arr1=np.array(list(range(1,1000000)))
arr2=np.array(list(range(1,1000000)))
arr1*arr2
end=time.time()
print(f"time taken {end-start}")
```
